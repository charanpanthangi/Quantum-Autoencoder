# Quantum Autoencoder (QAE) demo

This notebook walks through a tiny quantum autoencoder that compresses two-qubit states down to one qubit and then reconstructs them.

## Why use SVG plots?
GitHub renders SVG images directly in pull requests and keeps them easy to diff. We configure the notebook to output SVG so no binary images are created.

```python
%config InlineBackend.figure_formats = ['svg']
import matplotlib.pyplot as plt
import numpy as np

from app.dataset import generate_training_states
from app.plots import plot_latent_space, plot_reconstruction_loss
from app.trainer import evaluate_qae, train_qae
```

## Generate a few training states
We encode small angle pairs into two qubits. Keeping the dataset tiny makes the demo fast.

```python
training_states = generate_training_states(6)
training_states
```

## Train the quantum autoencoder
A short training run is enough to show the loss decreasing.

```python
results = train_qae(training_states, n_epochs=10, lr=0.2, seed=0)
loss_history = results['loss_history']
loss_history
```

## Plot the reconstruction loss
The curve should trend downward, showing the circuit is learning.

```python
plot_reconstruction_loss(loss_history, 'examples/qae_reconstruction_loss.svg')
from IPython.display import SVG
SVG(filename='examples/qae_reconstruction_loss.svg')
```

## Evaluate fidelity and latent space
We check how close reconstructions are to the original states and how often the trash qubit is near |0>.

```python
metrics = evaluate_qae(training_states, results['encoder'], results['decoder'])
metrics
```

```python
plot_latent_space(metrics['latent_probs'], [f'state {i}' for i in range(len(training_states))], 'examples/qae_latent_space.svg')
SVG(filename='examples/qae_latent_space.svg')
```
